## Cleaning And Transforming Data

```python
df["salary_range"] = df["salary_max_usd"] - df["salary_min_usd"] 
```

```python
df["is_senior"] = df["is_senior"].astype("bool")
```

```python
df["is_llm_role"] = df["is_llm_role"].astype("bool")
```

```python
df["is_remote_friendly"] = df["is_remote_friendly"].astype("bool")
```

```python
df["annual_salary_usd"] = df["annual_salary_usd"].astype("int")
```

```python
df["average_salary"] = (df["salary_min_usd"] + df["salary_max_usd"]) / 2
```

```python
df["skill_count"] = df["required_skills"].str.split(",").str.len()
```

```python
df["experience_group"] = pd.cut(
    df["years_of_experience"],
    bins=[0,2,5,8,20],
    labels=["Entry","Junior","Mid","Senior"]
)
```

## Data visualization

## 1. Which job categories offer the highest average salaries?

```python
avg_salary = (
    df.groupby("job_category", as_index=False)["annual_salary_usd"]
      .mean()
      .sort_values(by="annual_salary_usd", ascending=False))
plt.figure(figsize=(20,10))
sns.barplot(
    x="job_category",
    y="annual_salary_usd",
    data=avg_salary,
    hue="job_category",
    palette="viridis",
    legend=False)
plt.title("Average Salary by Job Category")
plt.xlabel("Job Category")
plt.ylabel("Average Salary (USD)")
plt.show()
```

## 2. Which cities have the highest-paying AI jobs?

```python
top_cities = (
    df.groupby("city", as_index=False)["annual_salary_usd"]
      .mean()
      .sort_values(by="annual_salary_usd", ascending=False)
      .head(10))
plt.figure(figsize=(12,5))
sns.barplot(
    x="city",
    y="annual_salary_usd",
    data=top_cities,
    hue="city",
    palette="Set2",
    legend=False)
plt.title("Top 10 Highest Paying Cities")
plt.xlabel("City")
plt.ylabel("Average Salary (USD)")
plt.show() 
```

## 3. Which countries have the highest-paying AI jobs?

```python
country_salary = (df.groupby("country", as_index=False)["annual_salary_usd"]
      .mean()
      .sort_values(by="annual_salary_usd", ascending=False))
plt.figure(figsize=(15,10))
sns.barplot(
    x="country",
    y="annual_salary_usd",
    data=country_salary,
    hue="country",
    palette="magma",
    legend=False)
plt.title("Average Salary by Country")
plt.xlabel("Country")
plt.ylabel("Average Salary (USD)")
plt.show() 
```

## 4. How does experience level affect annual salary?

```python
plt.figure(figsize=(10,5))
sns.boxplot(
    x="experience_level",
    y="annual_salary_usd",
    data=df,
    hue="experience_level",
    palette="coolwarm",
    legend=False)
plt.title("Salary by Experience Level")
plt.xlabel("Experience Level")
plt.ylabel("Annual Salary (USD)")
plt.show()
```

## 5. Do remote-friendly jobs pay more?

```python
plt.figure(figsize=(7,5))
sns.boxplot(
    x="is_remote_friendly",
    y="annual_salary_usd",
    data=df,
    hue="is_remote_friendly",
    palette="Set1",
    legend=False)
plt.title("Remote-Friendly vs Non-Remote Salary")
plt.xlabel("Remote Friendly")
plt.ylabel("Annual Salary (USD)")
plt.show() 
```

## 6. Which industries have the highest demand scores?

```python
industry_demand = (
    df.groupby("industry", as_index=False)["demand_score"]
      .mean()
      .sort_values(by="demand_score", ascending=False))
plt.figure(figsize=(20,10))
sns.barplot(
    x="industry",
    y="demand_score",
    data=industry_demand,
    hue="industry",
    palette="rocket",
    legend=False)
plt.title("Average Demand Score by Industry")
plt.xlabel("Industry")
plt.ylabel("Demand Score")
plt.show()
```

## 7. Do LLM roles receive higher salaries?

```python
plt.figure(figsize=(7,5))
sns.boxplot(
    x="is_llm_role",
    y="annual_salary_usd",
    data=df,
    hue="is_llm_role",
    palette="pastel",
    legend=False)
plt.title("Salary: LLM Roles vs Non-LLM Roles")
plt.xlabel("LLM Role")
plt.ylabel("Annual Salary (USD)")
plt.show()
```

## 8. Relationship between experience and salary

```python
plt.figure(figsize=(8,6))
sns.scatterplot(
    x="years_of_experience",
    y="annual_salary_usd",
    data=df,
    hue="experience_level",
    palette="viridis",
    s=80)
plt.title("Years of Experience vs Salary")
plt.xlabel("Years of Experience")
plt.ylabel("Annual Salary (USD)")
plt.show() 
```

## 9. AI Salary Premium vs Demand Score

```python
plt.figure(figsize=(8,6))
sns.scatterplot(
    x="ai_salary_premium_pct",
    y="demand_score",
    data=df,
    hue="salary_tier",
    palette="tab10",
    s=80)
plt.title("AI Salary Premium vs Demand Score")
plt.xlabel("AI Salary Premium (%)")
plt.ylabel("Demand Score")
plt.show() 
```

## 10. Which company sizes offer the best salaries and benefits? 

```python
company = (
    df.groupby("company_size", as_index=False)[["annual_salary_usd","benefits_score_10"]]
      .mean())
company = company.melt(
    id_vars="company_size",
    value_vars=["annual_salary_usd","benefits_score_10"],
    var_name="Metric",
    value_name="Value")
plt.figure(figsize=(10,5))
sns.barplot(
    x="company_size",
    y="Value",
    hue="Metric",
    data=company,
    palette="Set2")
plt.title("Salary and Benefits by Company Size")
plt.xlabel("Company Size")
plt.ylabel("Average Value")
plt.show()
```
