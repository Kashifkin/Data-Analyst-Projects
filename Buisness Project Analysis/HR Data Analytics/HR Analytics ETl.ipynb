### 1. Basic Statistics — Age 

```python
age = df['Age'].to_numpy()
income = df['MonthlyIncome'].to_numpy()
daily_rate = df['DailyRate'].to_numpy()
experience = df['TotalExperience(Years)'].to_numpy()
companies = df['NumCompaniesWorked'].to_numpy()

print("Mean Age:", np.mean(age))
print("Median Age:", np.median(age))
print("Minimum Age:", np.min(age))
print("Maximum Age:", np.max(age))
print("Standard Deviation:", np.std(age))
```

### 2. Check Missing Values 

```python
print("Missing Age:", np.sum(np.isnan(age)))
print("Missing Income:", np.sum(np.isnan(income))) 
```

##  3. Monthly Income Analysis 

```python
print("Mean Income:", np.mean(income))
print("Median Income:", np.median(income))
print("Minimum Income:", np.min(income))
print("Maximum Income:", np.max(income)) 
```

### 4. Experience Analysis 

```python
print("Average Experience:", np.mean(experience))
print("Minimum Experience:", np.min(experience))
print("Maximum Experience:", np.max(experience)) 
```

### 5. Find Employees with High Income

```python
high_income = income[income > np.mean(income)]
print("Number of employees earning above average:",len(high_income)) 
```

### 6. Find Possible Outliers in Monthly Income 

```python
Q1 = np.percentile(income, 25)
Q3 = np.percentile(income, 75)

IQR = Q3 - Q1

lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR

outliers = income[(income < lower_limit) | (income > upper_limit)]

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Number of outliers:", len(outliers)) 
```

### 7. NumPy Conditional Analysis 

```python
older_employees = age[age > 40]
print("Employees older than 40:", len(older_employees))
experienced = experience[experience > 5]
print("Employees with more than 5 years experience:",len(experienced)) 
```

### 8. Correlation Between Age and Income 

```python
correlation = np.corrcoef(age, income)[0, 1]
print("Correlation between Age and Income:", correlation)
```

## Cleaning and Transforming Data 

```python
df["YearsWithCurrManager"] = df["YearsWithCurrManager"].fillna("0")
```

```python
df.columns
```

```python
def experience_group(years):
    if years < 2:
        return "Fresher"
    elif years < 5:
        return "Junior"
    elif years < 10:
        return "Mid-Level"
    elif years < 20:
        return "Senior"
    else:
        return "Expert"
df["ExperienceGroup"] = df["TotalExperience(Years)"].apply(experience_group)
```

```python
def age_category(age):
    if age < 25:
        return "Young"
    elif age < 35:
        return "Adult"
    elif age < 50:
        return "Middle Age"
    else:
        return "Senior"

df["AgeCategory"] = df["Age"].apply(age_category) 
```
