## Cleaning And Transforming Data

```python
df["Returns"] = df["Returns"].fillna(0)
```

```python
df = df.drop(columns=["ind1", "ind2"])
```

```python
df["Sales"] = df["Sales"].astype('int')
```

```python
df["Returns"] = df["Returns"].astype('int')
```

```python
df.rename(columns={
    "Row ID+O6G3A1:R6": "Row_ID_O6G3A1_R6",
    "Order ID": "Order_ID",
    "Order Date": "Order_Date",
    "Ship Date": "Ship_Date",
    "Ship Mode": "Ship_Mode",
    "Customer ID": "Customer_ID",
    "Customer Name": "Customer_Name",
    "Sub-Category": "Sub_Category",
    "Payment Mode": "Payment_Mode"}, inplace=True)
```

```python
df.columns
```

## Business Questions

## 1. Highest Sales and Profit by Category

```python
Highest_Sales_by_Category = df.groupby("Category")["Sales"].sum()
Highest_Sales_by_Category
```

# 2. Sales and Profit by State

```python
sales_profit_by_state = df.groupby("State")[["Sales", "Profit"]].sum()
sales_profit_by_state.head()
```

# 3. Sales and Profit by City

```python
sales_profit_by_city = df.groupby("City")[["Sales", "Profit"]].sum()
sales_profit_by_city.head()
```

# 4. Sales and Profit by Customer Segment

```python
sales_profit_by_segment = df.groupby("Segment")[["Sales", "Profit"]].sum()
sales_profit_by_segment
```

# 5. Monthly Sales Trend

```python
df["Order_Date"] = pd.to_datetime(df["Order_Date"], format="%d-%m-%Y") 
monthly_sales = df.groupby(df["Order_Date"].dt.to_period("M"))["Sales"].sum()
monthly_sales.head()
```

# 6. Top Products by Sales and Profit

```python
sales_profit_by_product = df.groupby("Product Name")[["Sales", "Profit"]].sum() 
sales_profit_by_product.head()
```

# 7. Sales and Profit by Shipping Mode 

```python
sales_profit_by_ship_mode = df.groupby("Ship_Mode")[["Sales", "Profit"]].sum()
sales_profit_by_ship_mode
```

# 8. Sales and Quantity by Payment Mode

```python
sales_quantity_by_payment_mode = df.groupby("Payment_Mode")[["Sales", "Quantity"]].sum() 
sales_quantity_by_payment_mode
```

# 9. Return Analysis by Category

```python
returns_by_category = df.groupby("Category")[["Returns", "Profit"]].sum().round() 
returns_by_category
```

# 10. Top Customers by Sales and Profit

```python
sales_profit_by_customer = df.groupby("Customer_Name")[["Sales", "Profit"]].sum() 
sales_profit_by_customer.head()
```
