## Cleaning

```python
df["Purchase Amount (USD)"] = df["Purchase Amount (USD)"].fillna(df["Purchase Amount (USD)"].mean())
```

```python
df["Review Rating"] = df["Review Rating"].fillna(df["Review Rating"].mean())
```

```python
df["Previous Purchases"] = df["Previous Purchases"].fillna(df["Previous Purchases"].median())
```

```python
df["size"] = df["size"].fillna("Unknown")
```

```python
df.isnull().sum()
```

```python
df.head()
```

```python
df["Previous Purchases"] = df["Previous Purchases"].astype(int)
```

```python
df["Purchase Amount (USD)"] = df["Purchase Amount (USD)"].astype(int)
```

```python
df.head()
```

```python
df.columns = df.columns.str.replace(" ","_")
df.columns = df.columns.str.lower()
```

```python
df.columns
```

```python
df["review_rating"] = df["review_rating"].round(1)
```

```python
df.to_csv("Customer_Behavior.csv",index = False)
```
